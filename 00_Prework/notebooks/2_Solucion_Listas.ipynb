#### **2. Listas**

¿Qué es una lista?
Una lista es un tipo de variable que puede almacenar varios valores a la vez. Es como una caja especial que puede contener más de un objeto dentro.  
Los elementos dentro de una lista pueden ser números, textos o incluso otras listas. Las listas son muy útiles cuando necesitas almacenar y trabajar con varios datos relacionados.

4. Agregar un elemento al final de la lista

```python
colores = ["rojo", "verde"]
colores.append("azul")

print(colores)

# las listas tienen un método que se llama append
# una_lista.append(un_elemento)  lo que hace es meter al final de la lista   una_lista   ese un_elemento
# fíjate en el punto entre una_lista y append
# ese punto aparecerá mucho en Python
# le estamos diciendo que tome una_lista y le haga algo, en este caso, un append a ESA una_lista
```

5. Inserta un elemento en una posición específica

```python
colores = ["rojo", "verde"]
colores.insert(1, "amarillo")

print(colores)

# insert es similar a append, pero inserta en el índice que le decimos el elemento que le decimos
# ¿a qué lista? Sí, a la que está antes del punto y de insert
```

6. Elimina un elemento de la lista

```python
frutas = ["manzana", "naranja", "kiwi"]
frutas.remove("naranja")

print(frutas)

# remove es un método de las listas para eliminar
# ojo, modifica la lista (igual que append también modifica la lista o insert, cada uno sus modificaciones)
```

10. Itera sobre los elementos de la lista

```python
colores = ["rojo", "verde", "azul"]

for color in colores:
    print(color)

# Aquí usamos un bucle for por primera vez. No te preocupes si no lo entiendes del todo,
# lo veremos en detalle en el notebook 6. De momento, fíjate en que el bucle
# recorre la lista y ejecuta el print para cada elemento
# aquí quiero que tome cada valor de la lista colores
# lo asocie a color
# entre en en bucle y lo imprima
# si te fijas, el bucle dará tres vueltas
```

11. Slicing: cómo acceder a una porción de la lista

```python
numeros = [10, 20, 30, 40, 50]

print(numeros[1:3])   # [20, 30]: del índice 1 al 2 (el 3 no está incluido)
print(numeros[:2])    # [10, 20]: desde el principio hasta el índice 1
print(numeros[3:])    # [40, 50]: desde el índice 3 hasta el final

# lista[inicio:fin]
# inicio: índice donde empieza el trozo (incluido)
# fin:    índice donde termina (NO incluido)
# si no pones inicio, empieza desde 0
# si no pones fin, llega hasta el final
```

12. Ordena y revierte una lista

```python
puntuaciones = [34, 87, 12, 65, 45]

puntuaciones.sort()     # ordena de menor a mayor (modifica la lista original)
print(puntuaciones)     # [12, 34, 45, 65, 87]

puntuaciones.reverse()  # da la vuelta al orden (modifica la lista original)
print(puntuaciones)     # [87, 65, 45, 34, 12]

# IMPORTANTE: sort() y reverse() modifican la lista original
# a diferencia de upper() en strings, estos métodos SÍ cambian la lista
```
