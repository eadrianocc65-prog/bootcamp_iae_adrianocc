#### **1. Variables**

¿Qué es una variable?
En programación, una variable es un lugar donde guardamos datos, como números o textos. Puedes pensar en una variable como una "caja" en la que colocamos algo. Podemos poner cualquier tipo de datos en esta caja, y luego podemos usar ese valor más tarde en el programa. Las variables son muy útiles porque te permiten almacenar información y manipularla según sea necesario.

Veamos algunos ejemplos de variables en Python. El objetivo de momento es soltarnos un poco con Python. Intenta leer el código, entender qué hace, ejecutarlo, variarlo... No te preocupes si no entiendes absolutamente todo, es normal, el objetivo es coger confianza en Python.

11. Conversión entre tipos de datos

```python
numero_texto = "42"
print(type(numero_texto))   # <class str>, es decir, texto, no número

numero = int(numero_texto)  # convertimos el texto a número entero
print(type(numero))         # <class int>
print(numero + 8)           # ahora sí podemos operar: 50

# int()   convierte a número entero
# float() convierte a número decimal
# str()   convierte a texto
```

```python
# Ejercicio 11.1: convertir número decimal a entero
precio = 21.45
precio = int(precio)
print(precio)
```

```python
precio = 19.99
precio_entero = int(precio)   # pierde los decimales
print(precio_entero)          # 19

nota = 8
nota_texto = str(nota)        # convierte el número a texto
print("Tu nota es: " + nota_texto)  # ahora podemos concatenarlo con texto

# ojo: sin str(), Python no puede concatenar texto + número
# "Tu nota es: " + 8  daría TypeError
```

```python
# Ejercicio 11.2: Crear una variable con un número (edad = 25)
# conviténdola a texto usando str() y concaténandola
# con otros textos usando +.

edad = 25
edad = str(edad)
print('Tengo' + ' '+ edad + ' ' 'años' ' '+ 'de' + ' ' + 'edad')
```

##### Solución Problemas Propuestos

**Problema 1:**

Crea dos variables a y b y asigna valores de tu elección. Luego, imprime la suma, resta, multiplicación y división de ambos.

```python
a = 8
b = 4

print("Suma:", a + b)
print("Resta:", a - b)
print("Multiplicación:", a * b)
print("División:", a / b)

# print SIEMPRE tiene que imprimir por pantalla lo que le paso dentro de los paréntesis (es para lo que se diseñó)
```

**Problema 2:**

Crea una variable precio y asigna un valor. Si el precio es mayor o igual a 100, imprime "Producto costoso". Si no, imprime "Producto económico".

```python
precio = 120

if precio >= 100:               # los dos puntos es porque comienza un bloque de código
    print("Producto costoso")   # ¡ojo a estos espacios en blanco iniciales! si se cumple el if, entro en el bloque
else:                           # si no, dos puntos, ¡bloque de código!
    print("Producto económico") # entonces imprimiría esto otro
```
