#### **3. Strings**

Un string es simplemente una secuencia de caracteres (letras, números, símbolos). En Python, los strings se escriben entre comillas (simples o dobles).  
Los strings nos permiten trabajar con texto y son fundamentales en muchos programas, como los que hacen tareas de entrada y salida de información.

A continuación tienes varios ejemplos de Strings, ejecuta las celdas para ver el resultado:

3. Convierte a mayúsculas

10. Divide un string en una lista

```python
texto = "Python es fácil"
palabras = texto.split()

print(palabras)  # Imprime ['Python', 'es', 'fácil']


# split es un método de strings que separa por el espacio en blanco, por defecto, y devuelve una lista
```

```python
# Ejercicio 10.1

texto = "Buenos días, María"
divide_texto = texto.split()

print(divide_texto)  # Devuelve una lista de los elementos de una variable entre corchetes, separados por una coma
```

11. F-strings: la forma moderna de insertar variables en texto

```python
nombre = "Ana"
edad = 28
ciudad = "Madrid"

# sin f-string: incómodo
print("Me llamo " + nombre + ", tengo " + str(edad) + " años y vivo en " + ciudad)

# con f-string: mucho más claro
print(f"Me llamo {nombre}, tengo {edad} años y vivo en {ciudad}")

# una f-string empieza con la letra f antes de las comillas
# dentro de las llaves {} puedes poner cualquier variable
# Python la sustituye por su valor automáticamente
```

```python
# Ejercicio 11.1

nombre = "Adriano"
ciudad = "Arcos de la Frontera"

print(f"Mi nombre es {nombre} y vivo en {ciudad}") 

# f debe ir seguida de comillas
```

```python
precio = 19.99
producto = "Portátil"

print(f"El {producto} cuesta {precio} euros")
print(f"Con IVA: {precio * 1.21:.2f} euros")  # :.2f = solo 2 decimales

# dentro de las llaves también puedes hacer operaciones directamente
# el :.2f es un formato opcional para controlar los decimales que se muestran
```

```python
# Ejercicio 11.2

precio = 20.00 
producto = "pack de conectores"

print(f"El {producto} cuesta {precio:.2f} €.")
print(f"La base imponible del pruducto es de {precio / 1.21:.2f} euros." )
```

12. Secuencias de escape: caracteres especiales en strings

```python
print("Primera línea\nSegunda línea")  # \n = nueva línea
print("Nombre:\tAna")                   # \t = tabulación (espacio largo)

# \n y \t son secuencias de escape
# la barra invertida \\ le dice a Python que el siguiente carácter tiene significado especial
# \n = new line (salto de línea)
# \t = tab (tabulación)
```

```python
# Ejercicio 12.1

print("Me llamo Adriano.\nEstudio Python") # # \n hace un salto de línea; los espacios que pongamos alrededor también se imprimen
print("\tVivo en Arcos de la Frontera" )      
```
