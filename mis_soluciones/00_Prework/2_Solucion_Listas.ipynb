#### **2. Listas**

¿Qué es una lista?
Una lista es un tipo de variable que puede almacenar varios valores a la vez. Es como una caja especial que puede contener más de un objeto dentro.  
Los elementos dentro de una lista pueden ser números, textos o incluso otras listas. Las listas son muy útiles cuando necesitas almacenar y trabajar con varios datos relacionados.

11. Slicing: cómo acceder a una porción de la lista

```python
numeros = [10, 20, 30, 40, 50]

print(numeros[1:3])   # [20, 30]: del índice 1 al 2 (el 3 no está incluido)
print(numeros[:2])    # [10, 20]: desde el principio hasta el índice 1
print(numeros[3:])    # [40, 50]: desde el índice 3 hasta el final

# lista[inicio:fin]
# inicio: índice donde empieza el trozo (incluido)
# fin:    índice donde termina (NO incluido)
# si no pones inicio, empieza desde 0
# si no pones fin, llega hasta el final
```

```python
# Ejercicio 11.1: Obtenemos un trozo de la lista usando slicing

objetos = ["libro", "cuaderno", "bolígrafo", "ordenador"]

print(objetos[2:3])  # Imprime sólo el índice 2: ['bolígrafo']
print(objetos[:3])  # Imprime desde el índice 0 hasta el índice 2: ['libro', 'cuaderno', 'bolígrafo']
print(objetos[2:]) # Imprime desde el índice 2 hasta el final: ['bolígrafo', 'ordenador']
```

12. Ordena y revierte una lista

```python
puntuaciones = [34, 87, 12, 65, 45]

puntuaciones.sort()     # ordena de menor a mayor (modifica la lista original)
print(puntuaciones)     # [12, 34, 45, 65, 87]

puntuaciones.reverse()  # da la vuelta al orden (modifica la lista original)
print(puntuaciones)     # [87, 65, 45, 34, 12]

# IMPORTANTE: sort() y reverse() modifican la lista original
# a diferencia de upper() en strings, estos métodos SÍ cambian la lista
```

```python
# Ejercicio 12.1: Ordenamos una lista de números usando el método sort y luego invertimos el orden con reverse

numeros = [5, 2, 8, 1, 9]
numeros.sort()  # Ordena la lista de menor a mayor

print(numeros)  # Imprime la lista en orden ascendente: [1, 2, 5, 8, 9]

numeros.reverse()  # Invierte el orden de la lista
print(numeros)  # Imprime la lista en orden descendente: [9, 8, 5, 2, 1]
```
